```python
import numpy as np
```

```python
m = 1000000

# intercept
x1 = np.ones(m)

# normal distributed data
x2 = np.random.normal(loc=3, scale=2, size=m)
x3 = np.random.normal(loc=-1, scale=2, size=m)

# combine
X = np.column_stack((x1, x2, x3))

print(X.shape)
```

```python
# theta
theta = [[3], [1], [2]]

epsilon = np.random.normal(loc=0, scale=2**0.5, size=(m,1))

Y = X@theta + epsilon

print(Y.shape)
```

```python
# to get same indices
np.random.seed(42)

# shuffling indices
indices = np.random.permutation(m)

# shuffling actual values
X = X[indices]
Y = Y[indices]

# split index
split_index = int(0.8*m)

# divide data
X_train = X[:split_index]
Y_train = Y[:split_index]

X_val = X[split_index:]
Y_val = Y[split_index:]

print("X train shape: ", X_train.shape)
print("Y train shape: ", Y_train.shape)
print("X val shape: ", X_val.shape)
print("Y val shape: ", Y_val.shape)
```

```python
def mini_batch_gd(X_train, Y_train, theta, epochs, alpha, k):
    m_train = X_train.shape[0]

    for epoch in range(epochs):

        # we can shuffle train data here also
        # indices = np.permutation(m_train)
        # X_train = X_train(indices)
        # Y_train = Y_train(indices)

        for i in range(0, m_train, k):
            X_batch = X[i: i+k]
            Y_batch = Y[i: i+k]

            # batch size, because last batch could be less than k
            batch_size = X_batch.shape[0]

            # prediction
            h_theta = X_batch@theta

            # error
            error = h_theta-Y_batch

            # gradient
            gradients = (X_batch.T@error)/batch_size

            # update theta
            theta = theta - alpha*gradients


    return theta
        
```

```python
# theta we will predict
theta_p = np.zeros((3, 1))

for k in [1, 80, 8000, 800000]:
    theta_p = mini_batch_gd(X_train, Y_train, theta_p, 10, 0.001, k)
    print("for value of k: ", k)
    print("predicted theta:  ",theta_p)
    h_val = X_val@theta_p
    cost = np.sum((h_val - Y_val)**2)/(2*X_val.shape[0])
    print("cost: ", cost)
    
```
