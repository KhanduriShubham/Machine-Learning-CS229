```python
def costFunc(y_out, y_pred, m):
    return (0.5/m)*np.sum((y_pred - y_out)**2)
```

```python
def gradient_descent(x_in, y_out, w, epoch, alpha, m):
    finalCost = 0;
    for i in range(epoch):
        # predicting output
        h_theta = x_in @ w

        # calculating cost (j_theta)
        finalCost = costFunc(y_out, h_theta, m)

        # calculating gradients
        error = (h_theta - y_out)
        gradients = (x_in.T @ error)/m
        # print(gradients.shape)

        # updating weight
        assert w.shape == gradients.shape
        w = w - alpha*gradients

        
    return w, finalCost
```

```python
w, finalCost = gradient_descent(x_in, y_out, w, 100, 0.01, m)
```

```python
print("cost: ", finalCost)
print("weights: ", w[1:])
print("intercept:, ", w[0])
```
