```python
import numpy as np
```

```python
# read csv
X_features = np.loadtxt("logisticX(in).csv", delimiter = ",")
Y = np.loadtxt("logisticY(in).csv", delimiter = ',')
Y = Y.reshape(-1, 1)

# normalise features
mean = np.mean(X_features, axis=0)
std = np.std(X_features, axis=0)

X_norm = (X_features-mean)/std
```

```python
# adding intercept
m = X_norm.shape[0]

X = np.column_stack((np.ones(m), X_norm))
print(X.shape)
```

```python
# initialize theta
theta = np.zeros((3, 1))

# calculating sigmoid
z = X@theta
p = 1/(1+np.exp(-z))

print(p.shape)
print(p[:5])
```

```python
# gradient of log likelihood
gradients = X.T@(Y-p)

# weights hessian
weights = p*(1-p)
H = -X.T@(weights*X)

print(gradients.shape)
print(weights.shape)
print(H.shape)
```

```python
# calculating delta to update theta
delta = np.linalg.solve(H, gradients)

# update theta
theta = theta - delta

print(delta)
print(theta)
```

```python
# implementing the loop 
n_iter = 100


for i in range(n_iter):
    # calculating sigmoid
    z = X@theta
    p = 1/(1+np.exp(-z))

    # gradient
    gradients = X.T@(Y-p)

    # hessian calculation
    weights = p*(1-p)
    H = -X.T@(weights*X)

    # calculating delta to update theta
    delta = np.linalg.solve(H, gradients)

    # update theta
    theta = theta - delta
    
print(theta)
```

```python
np.linalg.norm(delta)
```

```python
# implementing using gradient descent
theta_gd = np.zeros((3, 1))
alpha = 0.01

for i in range(n_iter):
    # predicting
    z = X@theta_gd
    p = 1/(1+np.exp(-z))

    # gradient
    gradients = X.T@(p-Y)

    # update theta
    theta_gd = theta_gd - alpha*gradients

print(theta_gd)
```
